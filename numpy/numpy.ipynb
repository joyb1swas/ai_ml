# **Array indexing and Slicing**

1. Indexing and slicing of vector

1.   Indexing and slicing of matrix
2.   Boolean indexing

**Indexing slicing of vector**

```python
arr1 = np.arange(1,10,1)
arr1
```

```python
arr1[4]
```

```python
arr1[1:9:2]
```

**Indexing slicing of matrices**

```python
arr2 = np.arange(1,31,1).reshape(5,6)
arr2
```

```python
arr2[1] # only row accessing
```

```python
arr2[1:3] #only row slicing
```

```python
arr2[0:6,0:1] #only column accessing
```

```python
arr2[0:2,1:3] #accessing a particular portion
```

**Bool indexing**

```python
bool_index = arr2 %2 == 0
bool_index
arr3 = arr2[bool_index]
arr3
```
