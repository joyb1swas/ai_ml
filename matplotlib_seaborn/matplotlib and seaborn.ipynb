# **Importing libraries**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# **Inputs**

```python
x = np.linspace(0,5,111)
x
y = x**3
y
```

# **Axis and plot design**

```python
plt.plot(x,y)
plt.title("Random")
plt.xlabel("X axis")
plt.ylabel("Y axis")
```

# **Subplots**

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x**2,y)
plt.subplot(2,2,3)
plt.plot(x**2,x**3)
plt.subplot(2,2,4)
plt.plot(x,y)
```

# **object oriented way**
**1. add axes** ([left value,bottom value,height of the canvas,width of the canvas]) bold text

**2. Fixsize** (height,width)

**3. DPI** how clear the img wil be

```python
fig = plt.figure(figsize=(8,8), dpi=200)
axis1 = fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,0.3,0.3])
axis2.plot(x,y)
```

# **Save plot**

```python
fig = plt.figure(figsize = (1,1),dpi=200)
ax = fig.add_axes([0.1,0.1,6,6])
ax.plot(x,y)
plt.savefig("rndm.png")
```

**2. Hist plot** = it shows only one axis other axis form in a dot in the canvas.

**3. Box plot** = It shows in a box form.

**4. Distributed plot** = it is used to visualize how data values are distributed.


1.  dist plot(in dist plot we will use histplot ,kde means kernel density)
2.  joined plot
1.   pair plot(hue used to categories the data and palatte use to define the colour of the data points)

**5. categorical plot** = it is used ti plot categorical data.


1.   count plot
2.   Bar plot

```python
df = sns.load_dataset("tips")
df
```

# **Distributed**

```python
sns.histplot(df['total_bill'],kde=True)
```

```python
sns.jointplot(x="total_bill",y="tip",data=df, kind='reg')
```

```python
sns.pairplot(df,hue="sex",palette="rainbow")
```

# **Categorical**

```python
sns.countplot(x = df['sex'],hue=df['smoker'])
```

```python
sns.barplot(x=df['tip'], y=df['day'],estimator=np.sum )
```

```python
sns.boxplot(x=df["tip"],y= df["day"],palette="rainbow")
```

```python
sns.violinplot(x='day',y='total_bill',data = df)
```

```python
sns.stripplot(x='day',y='total_bill',data = df)
```

```python
sns.swarmplot(x='day',y='total_bill',data = df)
```

```python
sns.violinplot(x='tip',y='day',data=df)
sns.swarmplot(x='tip',y='day',data=df)
```

# **Matrix plot**

## **Heatmap**

```python
tipcor = df[["total_bill","tip","size"]]
tipcor
tipcor.corr() #correlation finding
sns.heatmap(tipcor.corr(),annot=True,cmap="rainbow")
```

## **Clustermap**

```python
sns.clustermap(tipcor.corr(),annot=True)
```

## **Pivot table Heat map**

```python
pvtflight = df1.pivot_table(values="passengers",index="month",columns="year")
pvtflight
```

```python
sns.heatmap(pvtflight)
```
