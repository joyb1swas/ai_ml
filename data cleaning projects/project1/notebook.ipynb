```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv("titanic.csv")
```

```python
dataset.info()
```

```python
df = dataset
df
```

```python
df = df.drop(columns='Cabin')
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.info()
```
